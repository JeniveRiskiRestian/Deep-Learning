```python
import torch
import torch.nn as nn

# 3 input -> output
model = nn.Linear(3, 1)
print(model)
```

```python
# Ambil data contoh (rumah pertama)
x = torch.tensor([[60.0, 2.0, 15.0]])    # [1, 3]

# Forward Pass
z = model(x)              # output linear
pred = torch.relu(z)      # aktivasi ReLU

print("Nilai z (sebelum aktivasi)", z)
print("Nilai prediksi (setelah ReLU)", pred)
```

```python
# Target harga rumah pertama (juta rupiah)
target = torch.tensor([[600.0]])    # bentuk [1, 1]

# Hitung loss (MSE)
criterion = nn.MSELoss()
loss = criterion(pred, target)

print("Prediksi:", pred.item())
print("Nilai target:", target.item())
print("Nilai Loss (MSE):", loss.item())
```

```python
# Reset gradient lama
model.zero_grad()

# Backward pass
loss.backward()

# Lihat gradient
print("Gradient weight (dL/dW):")
print(model.weight.grad)
print("Gradient bias (dL/db):")
print(model.bias.grad)
```

```python
optimizer = torch.optim.SGD(model.parameters(), lr=0.01)
print("Weight sebelum update:", model.weight.data)
print("Bias sebelum update:", model.bias.data)

# Update parameter
optimizer.step()

print("Weight setelah update:", model.weight.data)
print("Bias setelah update:", model.bias.data)
```
